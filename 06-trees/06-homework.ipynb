## Question 3

Experiment with the n_estimators parameter

- Try these values of this parameter: `[10, 50, 100, 150]`.
- Set `random_state` to `1`.
- Evaluate the model on the validation dataset.

```python
val_dicts = df_val.to_dict(orient='records')
n_dv = DictVectorizer(sparse=True)
# X_val = n_dv.fit_transform(val_dicts)
X_val = dv.transform(df_val.to_dict(orient='records'))

scores = []

for new_n_estimators in [10, 50, 100, 150]:
    model = RandomForestRegressor(n_estimators=new_n_estimators, random_state=1, n_jobs=-1)
    model.fit(X_train, y_train) # (X_val, y_val)

    y_pred = model.predict(X_val)
    
    score = rmse(y_val, y_pred)
    scores.append((new_n_estimators, score))
    
    print(f"new_n_estimators: {new_n_estimators}, rmse_out: {score}")
```

## Question 4

Select the best max_depth:

- Try different values of max_depth: `[10, 15, 20, 25]`
- For each of these values,
    - try different values of n_estimators from `[10, 50, 100, 150]`
    - calculate the mean RMSE
- Fix the random seed: `random_state=1`

```python
scores = []

for new_max_depth in [10, 15, 20, 25]:
    for new_n_estimators in [10, 50, 100, 150]:
        model = RandomForestRegressor(n_estimators=new_n_estimators, random_state=1, n_jobs=-1, max_depth=new_max_depth)
        model.fit(X_train, y_train) # (X_val, y_val)
    
        y_pred = model.predict(X_val)
        
        score = rmse(y_val, y_pred)
        scores.append((new_max_depth, new_n_estimators, score))
        
        #print(f"new_max_depth: {new_max_depth}, new_n_estimators: {new_n_estimators}, rmse_out: {score}")

df_scores = pd.DataFrame(scores, columns=['max_depth', 'n_estimators', 'rmse'])
df_scores.groupby('max_depth').rmse.mean().round(3)
```
