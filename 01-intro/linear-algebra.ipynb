## Vector operations

```python
import numpy as np
```

```python
u = np.array([2, 4, 5, 6])
2 * u
```

### Vector addition

```python
v = np.array([1, 0, 0, 2])
u + v
```

### Multiplication

```python
u * v
```

Dot product: multiply each element of u with the corresponding element of v, and then we sum the results

```python
u.dot(v)
```

Matrix-vector multiplication

```python
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1],
])
```

```python
U.shape
```

```python
u.dot(v)
```

```python
U.dot(v)
```

```python
u * v
```

```python
v2 = np.array([1, 0.5, 2, 1])
```

```python
U.dot(v2)
```

Matrix-matrix multiplication

```python
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1],
    [0, 2, 1],
    [2, 1, 0],
])
```

```python
U.dot(V)
```

### Identity matrix
The identity matrix is useful for explaining what a matrix inverse is

```python
I = np.eye(3)
```

```python
I
```

```python
V.dot(I)
```
