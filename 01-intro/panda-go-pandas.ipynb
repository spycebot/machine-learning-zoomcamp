```python
import numpy as np
import pandas as pd
```

The data is a list of lists - each sublist is a row in the table, so each row is a car - and a separate variable that defines the column names

```python
pd.__version__
```

```python
data = [
    ['Nissan', 'Stanza', 1991, 138, 4, 'MANUAL', 'sedan', 2000],
    ['Hyundai', 'Sonata', 2017, None, 4, 'AUTOMATIC', 'Sedan', 27150],
    ['Lotus', 'Elise', 2010, 218, 4, 'MANUAL', 'convertible', 54990],
    ['GMC', 'Acadia',  2017, 194, 4, 'AUTOMATIC', '4dr SUV', 34450],
    ['Nissan', 'Frontier', 2017, 261, 6, 'MANUAL', 'Pickup', 32340],
]

columns = [
    'Make', 'Model', 'Year', 'Engine HP', 'Engine Cylinders',
    'Transmission Type', 'Vehicle_Style', 'MSRP'
]
```

Now we can turn it into a DataFrame using the DataFrame constructor. 

```python
df = pd.DataFrame(data, columns=columns)
```

### Filtering
Inner brackets returns a new boolien Series with False and True values. The outer brackets look at the values and returns a data frame containing only True values.

```python
df[
    df['Year'] >= 2015
]
```

```python
df[
    df['Make'] == 'Nissan'
    ]
```

Combining conditions:

```python
df[
    (df['Make'] == 'Nissan') & (df['Year'] >= 2015)
        ]
```
