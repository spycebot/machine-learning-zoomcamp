Q7 Sum of Weights

```python
# All cars from Asia
df[
    df["origin"] == "Asia"
]
```

```python
# Only columns vehicle_weight and model_year
vw_my = df[
    ["vehicle_weight", "model_year"]
]
print(vw_my)
```

```python
# First 7 values
vw_my.head(n=7)
```

```python
# Underlying NumPy array
X = vw_my.values
print(X)
```

```python
# Matrix-matrix multiplication between the transpose of X and X
# XTX = X.dot(X.T) # Wrong per thread in Slack
XTX = X.T.dot(X) # Solution offered by sSlack thread
print(XTX)
```

```python
# Inverse of XTX
XTX_inv = np.linalg.inv(XTX)
print(XTX_inv)
```

```python
# Checking determinant of XTX
np.linalg.det(XTX)
```

```python
np.linalg.pinv(XTX)
```
